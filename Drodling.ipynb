```python

a = 0                   # startverdi intervall
b = 1                   # sluttverdi intervall
noyaktighet = 0.0001    # angir hvor nøyaktig svaret skal være 

def f(x):
    return x**3 + x - 1

m = (a + b)/2             # finner midtpunkt i opprinnelig intervall

while abs(f(m)) >= noyaktighet:
    if f(a)*f(m) < 0:
        b = m
    else:
        a = m

    m = (a + b)/2

print("Løsningen på likningen er ", round (m, 3))


```

```python
p = 1.0E3                                   # rho vann kg/m**3
V = 100*(1.0E-2)**3                         # Volum i m**3
g = 9.81                                    # tyngdeakselerasjonen N/kg

oppdrift = p * V * g                        # formel

print(f"Oppdriften er {oppdrift:.3f} N")    #:.3 gir tre desimaler

```
